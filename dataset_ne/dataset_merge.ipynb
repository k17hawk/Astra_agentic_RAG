```python
import unicodedata
import heapq

def normalize_python(s, folding_table=None):
    s = unicodedata.normalize('NFC', s)
    if folding_table:
        s = ''.join(folding_table.get(c, c) for c in s)
    s = s.replace('\u200D', '')   # remove ZWJ, keep ZWNJ
    return s

def is_devanagari_word(w):
    return any('\u0900' <= c <= '\u097F' for c in w)

longest = []   # (length, word) heap, keeps the shortest among the top 5

with open("ne (2).txt", "r", encoding="utf-8") as f:
    for line in f:
        normalized_line = normalize_python(line)
        for word in normalized_line.split():
            if not is_devanagari_word(word):
                continue
            if len(longest) < 5:
                heapq.heappush(longest, (len(word), word))
            elif len(word) > longest[0][0]:
                heapq.heapreplace(longest, (len(word), word))

# Sort descending by length
longest_words = sorted(longest, key=lambda x: -x[0])

for i, (length, w) in enumerate(longest_words):
    print(f"\n{'='*60}")
    print(f"Longest Nepali word #{i+1} (len={length})")
    print(f"{'='*60}")
    print(f"First 150 chars:  {w[:150]!r}")
    print(f"Last 150 chars:   {w[-150:]!r}")

    dev = sum(1 for c in w if '\u0900' <= c <= '\u097F')
    lat = sum(1 for c in w if c.isascii() and c.isalpha())
    dig = sum(1 for c in w if c.isdigit())
    punc = sum(1 for c in w if not c.isalnum())
    other = len(w) - dev - lat - dig - punc

    print(f"\nBreakdown:")
    print(f"  Devanagari: {dev:6} ({100*dev/len(w):.1f}%)")
    print(f"  Latin:      {lat:6} ({100*lat/len(w):.1f}%)")
    print(f"  Digits:     {dig:6} ({100*dig/len(w):.1f}%)")
    print(f"  Punctuation:{punc:6} ({100*punc/len(w):.1f}%)")
    print(f"  Other:      {other:6} ({100*other/len(w):.1f}%)")
    
```

```python
import unicodedata

def normalize_python(s, folding_table=None):
    s = unicodedata.normalize('NFC', s)
    if folding_table:
        s = ''.join(folding_table.get(c, c) for c in s)
    s = s.replace('\u200D', '')   # remove zero-width joiner
    return s

# Characters to replace with a space (split them into separate words)
SEPARATORS = [',', '–']          # <-- comma and en dash (U+2013)
                                  # Add '-' if you also want to break on regular hyphens

LENGTH_THRESHOLD = 50
input_path = "ne3.txt"
output_path = "ne3_cleaned.txt"

with open(input_path, "r", encoding="utf-8") as fin, \
     open(output_path, "w", encoding="utf-8") as fout:
    for line in fin:
        normalized = normalize_python(line)
        # Replace all chosen separators with a space
        for sep in SEPARATORS:
            normalized = normalized.replace(sep, ' ')
        tokens = normalized.split()
        kept = [t for t in tokens if len(t) <= LENGTH_THRESHOLD]
        fout.write(' '.join(kept) + '\n')

print(f"Done. Cleaned file saved to {output_path}")
```

```python
import os

def extract_only_content(root_folder, output_file):
    """
    Extract ONLY the article content from all .txt files,
    removing ALL metadata, headers, folder names, etc.
    """
    
    txt_files = []
    
    # Walk through all subfolders
    for foldername, subfolders, filenames in os.walk(root_folder):
        for filename in filenames:
            if filename.endswith('.txt'):
                full_path = os.path.join(foldername, filename)
                txt_files.append(full_path)
    
    print(f"Found {len(txt_files)} .txt files")
    
    with open(output_file, 'w', encoding='utf-8') as outfile:
        article_count = 0
        
        for filepath in sorted(txt_files):
            try:
                with open(filepath, 'r', encoding='utf-8') as infile:
                    content = infile.read()
                    
                    # Split on the separator line (===)
                    if '===' in content:
                        # Get everything after the === separator
                        article_content = content.split('===', 1)[-1]
                        # Clean up any remaining = signs
                        article_content = article_content.strip('=\n\r ')
                    else:
                        article_content = content
                    
                    # Only write if there's actual content
                    if article_content.strip():
                        outfile.write(article_content.strip())
                        outfile.write('\n\n')
                        article_count += 1
                        
            except Exception as e:
                print(f"Error processing {filepath}: {e}")
    
    print(f"Done! Extracted {article_count} articles to: {output_file}")

# Run it
extract_only_content("English", "all_articles.txt")
```
