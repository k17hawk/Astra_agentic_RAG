```python
dataset = load_dataset("wikimedia/wikipedia", "20231101.ne")
# Save all Nepali Wikipedia articles to a text file
output_file = "nepali_wikipedia.txt"

with open(output_file, "w", encoding="utf-8") as f:
    for i, example in enumerate(dataset['train']):
        title = example.get('title', '')
        text = example.get('text', '')
        
        # Write article with title separator
        f.write(f"=== {title} ===\n")
        f.write(text + "\n\n")
        
        # Optional: print progress
        if i % 1000 == 0:
            print(f"Processed {i} articles...")

print(f"Saved to {output_file}")
```

```python
from pathlib import Path

# Base path where the dataset lives
base_path = Path.home() / ".cache/kagglehub/datasets/ffatty/plaintext-wikipedia-full-english/versions/1/fullEnglish"

wiki_files = list(base_path.rglob("wiki_*"))
print(f"Found {len(wiki_files)} files")

# Output file
output_path ="wikipedia_combined.txt"
print(f"Combining into: {output_path}")

with open(output_path, 'w', encoding='utf-8') as out:
    for i, file_path in enumerate(wiki_files, 1):
        print(f"Processing {i}/{len(wiki_files)}: {file_path.name}")
        with open(file_path, 'r', encoding='utf-8') as inp:
            for chunk in iter(lambda: inp.read(8192), ''):
                out.write(chunk)
        out.write("\n\n")   # add separator between files

print(f"✅ Done! Combined file saved to: {output_path}")
print(f"Final size: {output_path.stat().st_size / 1e9:.2f} GB")
```
